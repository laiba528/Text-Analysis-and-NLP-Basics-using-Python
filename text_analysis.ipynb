```python
text = """ Nature language processing is  a field of artificial intelligence
nlp help computer understand  human language.
Text preprocessing is an important step in NLP.NLP is very interesting and useful"""
print("original text \n",text)
from collections import Counter

#word & character counts
word =text.split()
word_count=len(word)
char_count=len(text)
# 1. Counts
words = text.split()
print(f"Total Words: {len(words)}")
print(f"Total Characters: {len(text)}")

# 2. Vowel Analysis
vowels = "aeiouAEIOU"
vowel_count = sum(1 for c in text if c in vowels)
print(f"Total Vowels: {vowel_count}")

# 3. Repeated Words
cleaned_text = text.lower().replace(".", "")
word_freq = Counter(cleaned_text.split())
repeated = {w:c for w,c in word_freq.items() if c > 1}

print(f"Word Frequency: {dict(word_freq)}")
print(f"Repeated Words: {repeated}")


```
